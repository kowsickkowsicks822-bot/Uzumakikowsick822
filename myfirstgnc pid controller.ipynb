```python
import matplotlib.pyplot as plt
import numpy as np

dt = 0.1
total_time = 10.0
steps = int(total_time/dt)

altitude = 0.0
velocity = 0.0
gravity = 9.81
thrust = 10.0

time_history = []
altitude_history = []

for i in range (steps):
       acceleration = thrust -gravity
       velocity += acceleration*dt
       altitude += velocity*dt

       if altitude <0:
        altitude = 0

       time_history.append(i*dt)
       altitude_history.append(altitude)


plt.plot(time_history,altitude_history)
plt.xlabel("Time (s)")
plt.ylabel("Altitude (m)")
plt.title("Open-loop Drone altitude Simulation")
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

dt = 0.1
total_time = 20.0
steps = int(total_time/dt)

altitude = 0.0
velocity = 0.0
gravity = 9.81
target_altitude = 50.0

kp = 2.0
ki = 0.0
kd = 0.0

integral_error = 0.0
previous_error = 0.0

time_history = []
altitude_history = []

for i in range (steps):
       error  = target_altitude-altitude

       thrust  = gravity+(kp*error)
       thrust = max(0.0,thrust)

       acceleration = thrust -gravity
       velocity += acceleration*dt
       altitude += velocity*dt

       if altitude <0:
        altitude = 0
        velocity = 0

       time_history.append(i*dt)
       altitude_history.append(altitude)


plt.plot(time_history,altitude_history)
plt.axhline(y=target_altitude,color="r",linestyle="--",label="Target")
plt.xlabel("Time (s)")
plt.ylabel("Altitude (m)")
plt.title("P-controller altitude Simulation")
plt.show()
```
