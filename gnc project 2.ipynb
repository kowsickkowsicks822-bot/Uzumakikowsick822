```python
import numpy as np
import matplotlib.pyplot as plt

altitude = 0
velocity =0
altitudes = []
errors =[]

mass = 1.5
gravity = 9.81
dt = 0.1
target_altitude =50

kp = 2.0
kd = 8

previous_error = 0

for t in range (200):
      error = target_altitude -altitude
      derivative = (error - previous_error)/dt
      thrust = kp *error + mass *gravity+kd*derivative
      weight = mass *gravity
      net_force = thrust - weight
      acceleration = net_force /mass
      velocity = velocity + acceleration * dt
      altitude = altitude + velocity * dt
      altitudes.append(altitude)
      errors.append(error)
      previous_error = error

fig,(ax1,ax2) = plt.subplots(2,1,figsize=(6,8))
ax1.plot(altitudes,label = "altitude")
ax1.axhline(y= target_altitude , color ='r',linestyle = '--' , label = "target")
ax1.set_title("Altitude vs target")
ax1.legend()

ax2.plot(errors, color = 'orange')
ax2.axhline(y= 0 , color ='g',linestyle = '--' )
ax2.set_title("Error over time")

plt.tight_layout
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

altitude = 0
velocity =0
altitudes = []
errors =[]

mass = 1.5
gravity = 9.81
dt = 0.1
target_altitude =50

kp = 6.0
kd = 5
ki = 0.1
integral = 0

previous_error = 0

for t in range (200):
      error = target_altitude -altitude
      derivative = (error - previous_error)/dt
      integral =integral+error*dt
      integral = max(-50,min(50,integral))
      thrust = kp *error + mass *gravity+kd*derivative+ki*integral
      weight = mass *gravity
      net_force = thrust - weight
      acceleration = net_force /mass
      velocity = velocity + acceleration * dt
      altitude = altitude + velocity * dt
      altitudes.append(altitude)
      errors.append(error)
      previous_error = error

fig,(ax1,ax2) = plt.subplots(2,1,figsize=(6,8))
ax1.plot(altitudes,label = "altitude")
ax1.axhline(y= target_altitude , color ='r',linestyle = '--' , label = "target")
ax1.set_title("Altitude vs target")
ax1.legend()

ax2.plot(errors, color = 'orange')
ax2.axhline(y= 0 , color ='g',linestyle = '--' )
ax2.set_title("Error over time")

plt.tight_layout
plt.show()
```
